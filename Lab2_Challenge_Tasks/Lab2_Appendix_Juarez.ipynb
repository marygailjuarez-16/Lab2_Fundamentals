```python
# [CELL 1] EXERCISE 1: SENSOR MONITORING SYSTEM
SURNAME = "Juarez"
SEED_NUM = 3

# 1. Generate Sensor Data
raw_readings = [74, SEED_NUM, 77, "ERROR", 0, 103]

valid_readings = []
invalid_count = 0
execution_log = []

# 2. Process and Validate Inputs
for reading in raw_readings:
    try:
        val = int(reading)
        if 0 <= val <= 100:
            valid_readings.append(val)
            # 3. Classify Readings
            if val <= 30:
                classification = "Low Range"
            elif val <= 80:
                classification = "Normal Operation"
            else:
                classification = "High Range"
            execution_log.append(f"Reading {val}: Valid ({classification})")
        else:
            invalid_count += 1
            execution_log.append(f"Reading {val}: Invalid (Out of Safety Range 0-100)")
    except ValueError:
        invalid_count += 1
        execution_log.append(f"Reading '{reading}': Invalid (Non-numeric Input)")

# 4. Display Summary
print("--- EXERCISE 1 SUMMARY ---")
print(f"Engineer: {SURNAME}")
print(f"Generated Data: {raw_readings}")
print(f"Valid Readings: {valid_readings}")
print(f"Execution Log:")
for log in execution_log:
    print(f" - {log}")
print(f"Final Status: Total={len(raw_readings)} | Valid={len(valid_readings)} | Invalid={invalid_count}")
```

```python
# [CELL 2] EXERCISE 2: SIGNAL ANALYSIS
SURNAME = "Juarez"
SEED_NUM = 3

# 1. Generate Raw Signal
raw_signal = f"  {SURNAME}-{SEED_NUM}!  "

# 2. Process and Normalize
processed_signal = raw_signal.strip().upper()

# 3. Analyze Character by Character
letters = 0
digits = 0
specials = 0

for char in processed_signal:
    if char.isalpha():
        letters += 1
    elif char.isdigit():
        digits += 1
    else:
        specials += 1

# 4. Display Results
print("--- EXERCISE 2 SUMMARY ---")
print(f"Generated Raw Signal: '{raw_signal}'")
print(f"Processed Signal: '{processed_signal}'")
print("Character Analysis Breakdown:")
print(f" - Uppercase Letters: {letters}")
print(f" - Digits: {digits}")
print(f" - Special Symbols/Delimiters: {specials}")
print(f"Signal Classification: Mixed Alphanumeric Signal")
print(f"Final Output: Analyzed {len(processed_signal)} characters successfully.")
```

```python
# [CELL 3] EXERCISE 3: ELECTRONIC LOCK SYSTEM
SURNAME = "Juarez"
SEED_NUM = 3

# 1. Generate Credentials & Set Limits
correct_password = f"{SURNAME}#{SEED_NUM}"
attempt_limit = 3
attempts_made = 0
access_granted = False
execution_log = []

# Simulated User Inputs (2 wrong attempts, then 1 correct attempt)
simulated_inputs = ["WrongPass1", "WrongPass2", correct_password]

# 2. Manage Authentication Loop
for user_input in simulated_inputs:
    attempts_made += 1
    if user_input == correct_password:
        access_granted = True
        execution_log.append(f"Attempt {attempts_made}: Correct password entered. ACCESS GRANTED.")
        break
    else:
        execution_log.append(f"Attempt {attempts_made}: Invalid password ('{user_input}'). ACCESS DENIED.")

# 3. Final State Determination
final_state = "UNLOCKED" if access_granted else "LOCKED / TERMINATED"

# 4. Display Summary
print("--- EXERCISE 3 SUMMARY ---")
print(f"Generated Password: {correct_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {'GRANTED' if access_granted else 'DENIED'}")
print(f"Final System State: {final_state}")
print("Execution Log:")
for log in execution_log:
    print(f" - {log}")
```
